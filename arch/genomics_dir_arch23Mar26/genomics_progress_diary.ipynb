#### 2Mar26

observation: in db2_genomics_in when identifying genomics data-record date relationships, some (240) patients have 'genomics after record' but they have only a single record. How is this possible? Most likely explanation, supported by spot checks although not formally tested, is the following: 
- in the tte table, the final_date is NOT the same as the medical_record_closing_date, but it's the date of the last measurement
- this means that a record can be open even after the last measurement, 
- because either the patient drops out or disappears, or simply data collection stops or data is lost. 
- in these patients (this is what spot checks show), they start intervention, order genomics, their results arrive and the date is recorded, 
- BUT their last measurement is recorded before the arrival of their genomics test result, because of dropout or data loss.
- There are even cases where data is collected for hundreds of days, nearly two years, and the genomics result arrives in the last few months or weeks, and data collection stops before that. 

Data from these patients can be useful in genetics-to-obesity associations, and to some level, even genetics to weight loss. 

Actually, great idea - we can study the effects of genetics on weight loss WITHOUT PERSONALIZATION on records that at a later point, get genomics, but they also have non-personalized records!!



observation: other patients have genomics BEFORE record tag in their first, sometimes only, record (n 119 first, 97 first and only). How can this be possible? 

According to spot checks, these patients had a record opened, with no measurements for a period of some days/weeks/months, maybe getting a genomics result in that time. This reinforces the observation that medical record open/close dates do not coincide with measurement dates - which makes sense as one is done by patients and the other by doctors.
